# Problem 1

Kumanda isminde bir sınıf oluşturarak aşağıdaki işlemleri yapmasını sağlayan fonksiyonları sınıf içerisinde yazalım.
1. Tv Aç
2. Tv Kapat
3. Ses Ayarları
4. Kanal Ekle
5. Kanal Sayısını Öğrenme
6. Rastgele Kanala Geçme
7. Televizyon Bilgileri


Kumandayı kapatmak için q tuşuna basılmalı ve diğer durumlarda hangi işlemi yapmak istediği sorulmalıdır. Her işlemde gerekli fonksiyonunun sınıf içerisinden çağırılması gerekmektedir.

random ve time kütüphanesini araştırarak kullanabilirsiniz.

Kendi istediğiniz özellikleri eklemekte serbetsiniz.

```python
import random

class Kumanda:
    def __init__(self):
        self.tv_acik = False
        self.ses = 12
        self.kanallar = []

    def tv_ac(self):
        self.tv_acik = True
        print("TV açildi.")

    def tv_kapat(self):
        self.tv_acik = False
        print("TV kapatildi.")

    def ses_ayarlari(self):
        if self.tv_acik:
            yeni_ses = int(input("Yeni ses seviyesini giriniz (0-100): "))
            if 0 <= yeni_ses <= 100:
                self.ses = yeni_ses
                print(f"Ses seviyesi {self.ses}.")
            else:
                print("Ses seviyesi 0 ile 100 arasinda olmali.")
        else:
            print("TV açik değil.")

    def kanal_ekle(self):
        if self.tv_acik:
            kanal = input("Eklemek istediğiniz kanali girin: ")
            self.kanallar.append(kanal)
            print(f"{kanal} kanali eklendi.")
        else:
            print("TV acik değil.")

    def kanal_sayisini_ogren(self):
        if self.tv_acik:
            print(f"Toplam {len(self.kanallar)} kanal var.")
        else:
            print("TV acik değil.")

    def rastgele_kanala_gec(self):
        if self.tv_acik:
            if self.kanallar:
                kanal = random.choice(self.kanallar)
                print(f"Rastgele seçilen kanal: {kanal}")
            else:
                print("Kanal listesi boş.")
        else:
            print("TV acik değil.")

    def tv_bilgileri(self):
        if self.tv_acik:
            print(f"TV Acik. Ses seviyesi: {self.ses}. Kanal sayisi: {len(self.kanallar)}")
        else:
            print("TV Kapali.")

def main():
    kumanda = Kumanda()
    
    while True:
        print("\nSeçim yapin:")
        print("1. TV Aç")
        print("2. TV Kapat")
        print("3. Ses Ayarla")
        print("4. Kanal Ekle")
        print("5. Kanal Sayisi")
        print("6. Rastgele Kanal")
        print("7. TV Bilgileri")
        print("q. Cikis")
        
        secim = input("Seciminizi girin: ")
        
        if secim == '1':
            kumanda.tv_ac()
        elif secim == '2':
            kumanda.tv_kapat()
        elif secim == '3':
            kumanda.ses_ayarlari()
        elif secim == '4':
            kumanda.kanal_ekle()
        elif secim == '5':
            kumanda.kanal_sayisini_ogren()
        elif secim == '6':
            kumanda.rastgele_kanala_gec()
        elif secim == '7':
            kumanda.tv_bilgileri()
        elif secim == 'q':
            print("Cikiliyor...")
            break
        else:
            print("Gecersiz secim, tekrar deneyin.")

if __name__ == "__main__":
    main()
```
